```python
def convert_stringTobinary(string):
    return ''.join(format(ord(x), 'b') for x in string)
```

```python
def encode_message(bits, bases):
    message = []
    for i in range(n):
        qc = QuantumCircuit(1,1)
        if bases[i] == 0: # Prepare qubit in Z-basis
            if bits[i] == 0:
                pass 
            else:
                qc.x(0)
        else: # Prepare qubit in X-basis
            if bits[i] == 0:
                qc.h(0)
            else:
                qc.x(0)
                qc.h(0)
        qc.barrier()
        message.append(qc)
    return message
```

```python
def measure_message(message, bases):
    backend = Aer.get_backend('aer_simulator')
    measurements = []
    for q in range(n):
        if bases[q] == 0: # measuring in Z-basis
            message[q].measure(0,0)
        if bases[q] == 1: # measuring in X-basis
            message[q].h(0)
            message[q].measure(0,0)
        aer_sim = Aer.get_backend('aer_simulator')
        qobj = assemble(message[q], shots=1, memory=True)
        result = aer_sim.run(qobj).result()
        measured_bit = int(result.get_memory()[0])
        measurements.append(measured_bit)
    return measurements
```

```python
def remove_garbage(a_bases, b_bases, bits):
    good_bits = []
    for q in range(n):
        if a_bases[q] == b_bases[q]:
            # If both used the same basis, add
            # this to the list of 'good' bits
            good_bits.append(bits[q])
    return good_bits
```

```python
def sample_bits(bits, selection):
    sample = []
    for i in selection:
        # use np.mod to make sure the
        # bit we sample is always in 
        # the list range
        i = np.mod(i, len(bits))
        # pop(i) removes the element of the
        # list at index 'i'
        sample.append(bits.pop(i))
    return sample
```

```python
#Jack sent a message to Rose.
string= input("enter a string for the message: \n")
    
np.random.seed(seed=0)
n = 100

## Step 1
# Jack generates bits
jack_bits = np.asarray(list(convert_stringTobinary(string))).astype('int')
n=len(jack_bits)

## Step 2
# Create an array to tell us which qubits
# are encoded in which bases
jack_bases = randint(2, size=n)
message = encode_message(jack_bits, jack_bases)

## Step 3
# Decide which basis to measure in:
Rose_bases = randint(2, size=n)
Rose_results = measure_message(message, Rose_bases)
## Step 4
jack_key = remove_garbage(jack_bases, Rose_bases, jack_bits)
Rose_key = remove_garbage(jack_bases, Rose_bases, Rose_results)

## Step 5
sample_size = 15
bit_selection = randint(n, size=sample_size)

Rose_sample = sample_bits(Rose_key, bit_selection)
print("  Rose_sample = " + str(Rose_sample))
jack_sample = sample_bits(jack_key, bit_selection)
print("jack_sample = "+ str(jack_sample))
```
